```python
from pathlib import Path, PosixPath

def get_files(csv_path: str) -> list:
    path_ = Path(csv_path)
    if not path_.exists():
        print(f"Directory '{csv_path}' does not exist.")
        return []
    files = list(path_.glob('*.csv'))
    if not files:
        print(f"No .csv files found in '{csv_path}'.")
    return files

def get_backend_ds_name(c: PosixPath) -> tuple:
    parts = c.name.split('__')
    ds_name = parts[0].replace("rows_", "")
    backend_name = parts[1]
    return ds_name, backend_name
```

```python
import pandas as pd

csv_files = get_files("lambda_grid_results/coarse")
for c in csv_files:
    ds_name, backend_name = get_backend_ds_name(c)
    print("-" * 50, ds_name, "-", backend_name, "-" * 50)
    df600 = pd.read_csv(c).head(600)
    print(df600.groupby(["lambda1", "lambda2", "lambda3"])[
        ["validity","prox_cont_mad_negmean","div_cont_mad","robustness_keep_rate"]
    ].mean().describe(), "\n")
```

```python
import numpy as np

# 1. Create a default random number generator
rng = np.random.default_rng()

# 2. Use the generator to create random numbers

# Generate 5 random floats between 0.0 and 1.0
random_floats = rng.random((5,5))
print(f"Random floats:\n{random_floats}\n")

# Generate a 2x3 array of random integers between 0 and 9 (exclusive of 10)
random_integers = rng.integers(low=0, high=10, size=(2, 3))
print(f"Random integers:\n{random_integers}\n")

# Generate 4 numbers from a standard normal distribution (mean=0, std=1)
normal_dist_numbers = rng.standard_normal(4)
print(f"Numbers from a normal distribution:\n{normal_dist_numbers}")
```
